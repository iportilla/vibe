## 1. Fibonacci Series — Simple vs. Efficient

**Task:** Generate Fibonacci numbers.

The Fibonacci sequence starts:

`0, 1, 1, 2, 3, 5, 8, 13, ...`

First, try a simple recursive solution.


```python
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


for i in range(10):
    print(fibonacci(i), end=" ")
```

### More efficient version

Instead of recalculating the same values many times, keep the previous two numbers.


```python
def fibonacci_fast(n):
    sequence = []
    a, b = 0, 1

    for _ in range(n):
        sequence.append(a)
        a, b = b, a + b

    return sequence


print(fibonacci_fast(10))
```

## 2. Find Duplicates — Nested Loops vs. Set

**Task:** Find duplicate values in a list.


```python
def find_duplicates(numbers):
    duplicates = []

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] == numbers[j]:
                if numbers[i] not in duplicates:
                    duplicates.append(numbers[i])

    return duplicates


numbers = [4, 7, 2, 4, 9, 7, 10]
print(find_duplicates(numbers))
```

### Improved version


```python
def find_duplicates_fast(numbers):
    seen = set()
    duplicates = set()

    for number in numbers:
        if number in seen:
            duplicates.add(number)
        seen.add(number)

    return list(duplicates)


print(find_duplicates_fast(numbers))
```

## 3. Search for a Number — Linear vs. Binary Search

**Task:** Determine whether a target value exists in a list.


```python
def search(numbers, target):
    for number in numbers:
        if number == target:
            return True

    return False


numbers = [3, 8, 12, 18, 25, 31, 42, 57]
print(search(numbers, 42))
```

### Binary search


```python
def binary_search(numbers, target):
    low = 0
    high = len(numbers) - 1

    while low <= high:
        mid = (low + high) // 2

        if numbers[mid] == target:
            return True
        elif numbers[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return False


print(binary_search(numbers, 42))
```

## 4. Find the Most Common Word

**Task:** Find the most frequently occurring word in some text.


```python
text = '''
AI is changing software development.
AI can generate code.
AI can also make mistakes.
'''

words = text.lower().split()

counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

most_common = max(counts, key=counts.get)

print("Most common word:", most_common)
print("Count:", counts[most_common])
```

## 5. Cybersecurity — Password Strength Checker

**Task:** Check whether a password satisfies some basic password rules.

> For this classroom exercise, use a made-up password rather than a real password.


```python
def check_password(password):

    if len(password) < 12:
        return "Weak: too short"

    if not any(c.isupper() for c in password):
        return "Weak: needs uppercase"

    if not any(c.islower() for c in password):
        return "Weak: needs lowercase"

    if not any(c.isdigit() for c in password):
        return "Weak: needs a number"

    return "Password passes basic checks"


password = "VibeCoding2026"
print(check_password(password))
```
